```python
fruits = ["apple","banana","mango","peach"]
print(fruits)
```

Accesing elements


```python
print(fruits[0])
print(fruits[-1])
```

List Methods

```python
fruits.append("orange") #add item 
print(fruits)
```

```python
fruits.insert(1,"grapes") #insert at index
print(fruits)
```

```python
fruits.remove("orange") #remove item of list
print(fruits)
```

```python
fruits.pop() #remove last item
print(fruits)
```

looping Through list

```python
for fruit in fruits:
    print(fruit)
```

Sets

```python
my_set = {1,2,3,4,4,5}
print(my_set)
```

Methods of set

```python
my_set.add(6)
my_set.remove(3)
print(my_set)
```

SET OPERATIONS

```python
A = {1,2,3}
B = {4,5,6}

print(A.union(B))
print(A.intersection(B))
print(A.difference(B))
```

Dictionary

```python
student = {
    "name": "Muneeb Haider",
    "class": "16",
    "course": "Data Science"
}

print(student)
    
```

Accessing elements

```python
print(student["name"])
print(student.get("class"))
```

Dictionaries Methods

```python
student["name"] = "Ali"
student["class"] = "17"
student.pop("course")
print(student)
```

looping through Dictionary

```python
for key, value in student.items():
    print(key, ":", value)
```
