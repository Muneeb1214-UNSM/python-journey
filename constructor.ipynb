```python
class Dog:
    def __init__(self):
        self.name = "buddy"
    def bark(self):
        print(f"{self.name} dog says woof")
```

```python
dog1 = Dog()
dog1.bark()
```

```python
class Car:
    def __init__(self,brand,color):
        self.brand = brand
        self.color = color
        
```

```python
car1 = Car("toyota","red")
print(car1.color)
print(car1.brand)
```

```python
car1.color="blue"
car2 = Car("honda", "white")
print(car2.brand)
print(car2.color)
```
