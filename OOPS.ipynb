```python
class Car:
    def __init__(self, color):
        self.color = color

    def start(self):
        print(f" {self.color} car started")


my_car = Car("red")
my_car.start()

my_bro_car = Car("blue")
my_bro_car.start()
```

```python
class Car:
    color = "red"
    def drive(self):
        print(f"{self.color} car is moving")
    def setcolor(self, new_color):
        self.color = new_color
```

```python
car3 = Car()
car3.setcolor("blue")
car3.drive()
```
